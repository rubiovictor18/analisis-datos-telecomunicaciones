#### **1. Factores Estacionales y Post-Temporada Festiva**  
   - **Efecto rebote post-Año Nuevo:** Es común que, tras festividades como Navidad y Año Nuevo, usuarios retomen actividades pendientes (ej: activación de números virtuales para uso personal o laboral).  
   - **Promociones de "Comienzo de Año":** Empresas suelen lanzar ofertas en enero para captar nuevos clientes (ej: descuentos en números virtuales, bonificaciones).  

#### **2. Eventos Externos o Cambios Regulatorios**  
   - **Nuevas regulaciones de telecomunicaciones:** Si hubo cambios legales (ej: requisitos de verificación más estrictos), usuarios podrían haber migrado masivamente a números virtuales para cumplir con normativas.  
   - **Lanzamiento de un servicio asociado:** Por ejemplo, integración con una plataforma popular (red social, app de mensajería) que requiriera números virtuales para registro.  

#### **3. Causas Técnicas o Operativas**  
   - **Corrección de un error previo:** Si hubo problemas técnicos en diciembre (ej: fallas en el sistema de registro), el 02/01 podría reflejar un "pico de recuperación" de solicitudes acumuladas.  
   - **Actualización de la base de datos:** Migraciones o limpiezas de registros antiguos que generaron recuentos artificialmente altos.  

#### **4. Comportamiento de Usuarios Masivos**  
   - **Clientes corporativos:** Empresas que adquirieron bloques de números para sus equipos al reiniciar operaciones en enero.  
   - **Uso fraudulento o especulativo:** Posible intento de acaparamiento para reventa (ej: números "premium" con cifras repetidas).

## Clasificación de Servicios por Ingreso Bruto:

#### Código 12

### Objetivos específicos:
- Agrupar servicios según su frecuencia de uso e ingresos generados.
- Justificar la frecuencia de análisis (diaria, semanal, mensual).
- Aplicar métricas de distancia (Euclidiana) para medir similitudes entre servicios.
- Interpretar los grupos resultantes y generar recomendaciones estratégicas.

**Metodología**

- Fuente: Archivo JSON (short-term.json) con 48,856 registros.
- Filtrado: Se consideraron solo servicios con estado "Completed" (29,745 registros).
- Tabla resumen: Se agruparon los servicios por:
  - Veces usado (frecuencia de consumo).
  - Ingresos totales (suma de ingresos brutos).


Se evaluaron tres frecuencias:
- Diaria: Máximo detalle, pero alto ruido en datos.
- Semanal (Elegida): Balance detalle-estabilidad, menor granularidad que diaria.
- Mensual: Estabilidad, pero pierde patrones semanales.

Justificación: La frecuencia semanal ofrece el mejor equilibrio para detectar patrones sin el ruido de las fluctuaciones diarias.

**Métricas de Distancia (Euclidiana)**
- Se calculó la distancia euclidiana normalizada entre pares de servicios para medir similitudes.
- Fórmula aplicada: Distancia = √((x2 − x1)² + (y2 − y1)²), donde x = veces usado, y = ingresos totales.

###Clustering con K-means
- Algoritmo: K-means (4 clusters).
- Preprocesamiento: Normalización con StandardScaler.
- Métrica de calidad: Silhouette Score (0.933, excelente separación).
